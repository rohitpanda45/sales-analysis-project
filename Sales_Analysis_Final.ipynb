This project analyses retail sales data to identify seasonal trends, top products, and predict sales using machine learning models.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('sales_data_final.csv')

print(df.shape)
print(df.columns.tolist())
print(df.head())
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Region'] = df['Region'].fillna(df['Region'].mode()[0])
df['Sales'] = df['Sales'].fillna(df['Sales'].median())

print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("New shape:", df.shape)
```

```python
df['Date'] = pd.to_datetime(df['Date'])

print(df.dtypes)
```

```python
df.describe()
```

```python
df.nunique()
```

```python
# Day 19 - Business Insights

# 1. Category revenue percentage
category_sales = df.groupby('Category')['Sales'].sum()
category_pct = (category_sales / df['Sales'].sum() * 100).sort_values(ascending=False)

# 2. Monthly average sales
monthly_avg = df.groupby(df['Date'].dt.to_period('M'))['Sales'].mean()
overall_avg = df['Sales'].mean()
peak_month = monthly_avg.idxmax()
peak_month_value = monthly_avg.max()
above_avg_pct = (peak_month_value / overall_avg - 1) * 100

# 3. Top and bottom 3 products
product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)

# 4. Monthly sales and R-squared
monthly_sales = df.groupby(df['Date'].dt.month)['Sales'].sum()

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

X = monthly_sales.index.values.reshape(-1, 1)
y = monthly_sales.values

model = LinearRegression()
model.fit(X, y)

r2 = r2_score(y, model.predict(X))

print("Category Revenue %:")
print(category_pct)

print("\nHighest Average Sales Month:")
print(peak_month, peak_month_value)
print("Above overall average:", round(above_avg_pct, 2), "%")

print("\nTop 3 Products:")
print(product_sales.head(3))

print("\nBottom 3 Products:")
print(product_sales.tail(3))

print("\nR-squared:", round(r2, 3))
```

## KEY FINDINGS

### Finding 1
**Finding:** Electronics generates 83.41% of total revenue, making it the dominant category.  
**Meaning:** The business depends heavily on Electronics for revenue generation.  
**Action:** Maintain strong inventory and marketing support for Electronics products.

### Finding 2
**Finding:** December has the highest average monthly sales at ₹199,761.79, which is 35.12% above the overall average.  
**Meaning:** December shows the strongest sales performance and may indicate increased customer demand.  
**Action:** Increase inventory and promotional activities before December.

### Finding 3
**Finding:** The top 3 products by sales are Laptop (₹100.35M), Tablet (₹26.13M), and Monitor (₹21.50M).  
**Meaning:** These products contribute the most to overall sales revenue.  
**Action:** Prioritize stock availability and marketing for these high-performing products.

### Finding 4
**Finding:** The bottom 3 products are Headphones (₹4.13M), Keyboard (₹3.61M), and Mouse (₹2.06M).  
**Meaning:** These products generate considerably less revenue than the top-performing products.  
**Action:** Review their pricing, promotions, and customer demand before increasing inventory.

### Finding 5
**Finding:** The R-squared score for the seasonal sales pattern is 0.122.  
**Meaning:** The seasonal model explains only about 12.2% of the variation in sales, indicating a weak seasonal relationship.  
**Action:** The business should consider other factors such as product, quantity, price, and region when planning sales.

## RECOMMENDATIONS

1. Increase inventory and promotional campaigns for Electronics products, especially Laptop, Tablet, and Monitor, because they are the strongest revenue generators.

2. Prepare additional stock and marketing campaigns before December while reviewing the pricing and demand of low-performing products such as Headphones, Keyboard, and Mouse.
